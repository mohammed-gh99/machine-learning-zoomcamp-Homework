```python
n = len(df)
n_test = int(0.15*n)
n_val = int(0.20*n)
n_train = int(n-n_test-n_val)
```

```python
n , n_test, n_val , n_train , n_train+n_val+n_train
```

```python
idx = np.arange(n)
```

```python
np.random.seed(2)
np.random.shuffle(idx)
```

```python
df_train = df.iloc[idx[:n_train]]
df_val = df.iloc[idx[n_train:n_train+n_val]]
df_test = df.iloc[idx[n_train+n_val:]]
```

```python
df_train = df_train.reset_index(drop=True)
df_val = df_val.reset_index(drop=True)
df_test = df_test.reset_index(drop=True)
```

```python
y_train = np.log1p(df_train.medhouseval.values)
y_val = np.log1p(df_val.medhouseval.values)
y_test = np.log1p(df_test.medhouseval.values)
```

```python
del df_train['medhouseval']
del df_val['medhouseval']
del df_test['medhouseval']
```

```python
def prepare_X(df):
    df = df.copy()
  
    df_num = df
    X = df_num.values
    return X
```

```python
def train_linear_regression_reg(X, y , r= 0.001):
    ones = np.ones(X.shape[0])
    X = np.column_stack([ones, X])

    XTX = X.T.dot(X)
    XTX = XTX + r * np.eye(XTX.shape[0])
    
    XTX_inv = np.linalg.inv(XTX)
    w_full = XTX_inv.dot(X.T).dot(y)

    return w_full[0], w_full[1:]
```

```python
def rmse(y,y_pred):
    error = y-y_pred
    se = error**2
    mse = se.mean()
    return np.sqrt(mse)
```

```python
for r in [0.0 , 0.00001 , 0.0001 , 0.001 , 0.1 , 1 , 10]:
    X_train = prepare_X(df_train)
    w0, w = train_linear_regression_reg(X_train, y_train , r=r)
    
    X_val = prepare_X(df_val)
    y_pred = w0 + X_val.dot(w)
    score = rmse(y_val, y_pred)
    print(r,w0,score)
```

```python
r = 0.001
X_train = prepare_X(df_train)
w0, w = train_linear_regression_reg(X_train, y_train , r=r)
    
X_val = prepare_X(df_val)
y_pred = w0 + X_val.dot(w)
score = rmse(y_val, y_pred)
score
```

```python
df_full_train = pd.concat([df_train, df_val])
```

```python
df_full_train = df_full_train.reset_index(drop=True)
df_full_train
```

```python
X_full_train = prepare_X(df_full_train)
```

```python
y_full_train = np.concatenate([y_train, y_val])
y_full_train
```

```python
w0, w = train_linear_regression_reg(X_full_train, y_full_train , r=0.001)
```

```python
X_test = prepare_X(df_test)
y_pred = w0 + X_test.dot(w)
score = rmse(y_test, y_pred)
score
```
